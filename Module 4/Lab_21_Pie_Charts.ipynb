### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
# Remove missing values
database_data = df['DatabaseWantToWorkWith'].dropna()

# Split multiple database responses separated by ;
database_list = database_data.str.split(';').explode()

# Remove extra spaces
database_list = database_list.str.strip()

# Count each database and get the top 5
top_5_databases = database_list.value_counts().head(5)

# Create pie chart
plt.figure(figsize=(8, 6))

plt.pie(
    top_5_databases,
    labels=top_5_databases.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Databases Respondents Want to Work With')
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
# Remove missing values
devtype_data = df['DevType'].dropna()

# Split multiple developer roles separated by ;
devtype_list = devtype_data.str.split(';').explode()

# Remove extra spaces
devtype_list = devtype_list.str.strip()

# Count each developer role and get the top 5
top_5_devtypes = devtype_list.value_counts().head(5)

# Create pie chart
plt.figure(figsize=(8, 6))

plt.pie(
    top_5_devtypes,
    labels=top_5_devtypes.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Developer Roles')
plt.show()
```
