### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here

df["Country"].unique()
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
country_mapping = {
    'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
    'United States of America': 'United States',
    'Republic of North Macedonia': 'North Macedonia',
    'Russian Federation': 'Russia',
    'United Republic of Tanzania': 'Tanzania',
    'Iran, Islamic Republic of...': 'Iran',
    'Venezuela, Bolivarian Republic of...': 'Venezuela',
    'Hong Kong (S.A.R.)': 'Hong Kong',
    'Democratic Republic of the Congo': 'DR Congo',
    'Congo, Republic of the...': 'Congo (Republic)',
    'Viet Nam': 'Vietnam',
    'Republic of Korea': 'South Korea',
    "Lao People's Democratic Republic": 'Laos',
    'Syrian Arab Republic': 'Syria',
    "Democratic People's Republic of Korea": 'North Korea',
    'Micronesia, Federated States of...': 'Micronesia',
    'Libyan Arab Jamahiriya': 'Libya',
    'Republic of Moldova': 'Moldova'
}

df['Country'] = df['Country'].replace(country_mapping)
display(df['Country'].unique())
```

```python
df["EdLevel"].unique()
```

```python
edlevel_mapping = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)": "Doctorate",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "Secondary School",
    "Primary/elementary school": "Primary School",
    "Something else": "Other"
}

df['EdLevel'] = df['EdLevel'].map(edlevel_mapping)
display(df['EdLevel'].unique())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here

employment_encoded = pd.get_dummies(df["Employment"], dtype=int)

employment_encoded.head(20)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here

df.isnull().sum().sort_values(ascending=False)

```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here

median_salary = df["ConvertedCompYearly"].median()

df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(median_salary)

df["ConvertedCompYearly"].isnull().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here

most_common = df["RemoteWork"].mode()[0]

df["RemoteWork"] = df["RemoteWork"].fillna(most_common)

df["RemoteWork"].isnull().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here

min_val = df["ConvertedCompYearly"].min()
max_val = df["ConvertedCompYearly"].max()

df["ConvertedCompYearly_Normalized"] = (df["ConvertedCompYearly"] - min_val) / (max_val - min_val)

df[["ConvertedCompYearly", "ConvertedCompYearly_Normalized"]]
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here

import numpy as np

df["ConvertedCompYearly_Log"] = np.log1p(df["ConvertedCompYearly"])

df["ConvertedCompYearly_Log"].head()
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df["YearsCodePro"].unique()
```

```python
yearcode_mapping = {'Less than 1 year': 0,
                'More than 50 years': 51}

df['YearsCodePro'] = df['YearsCodePro'].replace(yearcode_mapping)
display(df['YearsCodePro'].unique())
```

```python
## Write your code here

df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")

def experience_level(years):
    
    if pd.isnull(years):
        return "Unknown"
    
    elif years < 5:
        return "Beginner"
    
    elif years < 10:
        return "Intermediate"
    
    else:
        return "Experienced"

df["ExperienceLevel"] = df["YearsCodePro"].apply(experience_level)
```
